Read from CSV with manual schema

```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructField,StructType, IntegerType, StringType, DateType

spark = SparkSession.builder \
    .appName("MyApp") \
    .master("local[*]")\
    .getOrCreate()

empy_schema = StructType([StructField("emp_id",StringType(),True),
                          StructField("emp_name",StringType(),True),
                          StructField("email",StringType(),True),
                          StructField("dept_id",StringType(),True),
                          StructField("salary",IntegerType(),True),
                          StructField("hire_date",DateType(),True),
                          StructField("job_title",StringType(),True)])

"""Read from CSV with auto detect schema"""
empy_df = spark.read.format("csv")\
     .option("header","true")\
     .schema(empy_schema).load(r"/content/employee.csv")

empy_df = spark.read.format("csv")\
     .option("header","true")\
     .schema(empy_schema).load(r"/content/employee.csv")

dept_df = spark.read.option("header", "true").option("inferSchema", "true").csv("/content/department.csv")
```

# Q35. Find how many employees were hired each year.






```python
empy_df.createOrReplaceTempView("employees")


result = spark.sql("""
    SELECT count(emp_id),extract(year from hire_date)
    FROM employees
    GROUP BY extract(year from hire_date)
""")

empy_df.withColumn("hire_year", F.year(F.col("hire_date")))\
  .groupBy("hire_year")\
  .agg(F.count("*").alias("employee_count")) \
  .orderBy("hire_year") \
  .show()

```

# Q36. Find the year with the most hires.




```python
empy_df.withColumn("hire_year",F.year(F.col("hire_date"))) \
  .groupBy("hire_year") \
  .agg(F.count("emp_id").alias("emp_count"))\
  .orderBy(F.desc("emp_count"))\
  .limit(1)\
  .select("hire_year")\
  .show()
```

# Q37. Find average salary of employees hired before 2020.
